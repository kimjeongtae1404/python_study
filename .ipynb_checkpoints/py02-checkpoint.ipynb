# 문자열
## 문자열 포매팅

```python
print("i eat 3 apples.")
print("i eat %d apples." % 5)
print("i eat %f apples." %5.0)
print("i eat %s apples." %'six')
```

```python
#우정렬 좌정렬
print("ㅋㅋㅋㅋ%10s"  %"hi" )
print("%-10s ㅋㅋㅋㅋ"  %"hi" )
```

```python
#소수점 표현하기
print("ㅋㅋㅋ%10.3f" %3.141592)
```

### 1.2 format 함수를 이용한 포매팅  

```python
print(" i eat {0} of {1} apples." .format(3, 5))

total = 10 
ea = 3 

print(" i eat {0} of {1} apples." .format(ea, total))
```

```python
#우정렬
print( "{0:>10}".format("hi"))
print( "{0:<10}".format("hi"))
print( "{0:^10}".format("hi"))
print( "{0:=^10}".format("hi"))
print( "{0:-^10.4f}".format(3.141592))
print( "{0:^10b}".format(10))
```

### 1.3 f문자열 포매팅

```python
name ="홍길동"
age= 23
print(f"나의 이름은 {name:^10}이고, 나이는 {age}입니다.")
print(f"나의 이름은 {name:*<10}이고, 나이는 {age+1}입니다.")
print(f"나의 이름은 {name:=^11}이고, 나이는 {age+1}입니다.")
```

```python
# str의 find() 메서드는 문자열에서 특정 부분 문자열을 찾고, 그 위치를 반환합니다. 만약 찾는 문자열이 없으면 -1을 반환합니다.

ss= "나는 오늘 하루종일 배가 고프다"

print( ss.find("오늘"))
print( ss.find("내일"))



```

```python
#str의 index() 메서드는 문자열에서 특정 부분 문자열을 찾고, 그 위치를 반환합니다. 만약 찾는 문자열이 없으면 ValueError를 발생시킵니다.

ss= "나는 오늘 하루종일 배가 고프다"
print( ss.index("오늘"))
#print( ss.index("내일"))
```

```python
#lower, upper (소문자, 대문자 변환)

lulu= "my Life is Beautiful"
uu= lulu.upper()
print(uu)
ll=uu.lower()
print(ll)
```

```python
# counter

lulu.count('i')
```

```python
#lstrip, rstrip, strip (왼쪽 공백 제거, 오른쪽 공백 제거, 양쪽 공백 제거)

name = '베트맨'
test= f"{name:^10}"

print(test)
print(test.lstrip())
print(test.rstrip())
print(test.strip())
```

```python
#자주쓰는 메소드() [join, split, replace]

ex1= 'abcdefghijk'
ex2= ex1.split()
print(ex2)

ex2_join="-".join(ex1)
print(ex2_join)

ex2_split=ex2_join.split('-')
print(ex2_split)

ex3= ex1.replace('d', 'D')
print(ex3)
```

```python
## 문자열의 리스트
ss1=["python","apple","string","cycle"]
ss2=[]
for i in range(4) :
    for j in range(5) :

        print(ss1[i][j])

        ss2.append(ss1[i][j])



n=3

ss2 = [ss2[i * n:(i + 1) * n] for i in range((len(ss2) - 1 + n) // n )] 



print(ss2)


```

```python
"""
문제1)
바구니에 ['호박','감자','소주','맥주','피자'] 물품이 있습니다
꺼내고 싶은 물품을 입력하세요: 소주

바구니에 소주를 꺼내어 ['호박','감자','맥주','피자']가 남았습니다.


문제2)
추가할 물건 이름을 적어주세요. 치즈
어느 물품의 뒤에 넣어둘까요 ? 맥주
['호박','감자','맥주','치즈','피자']
"""
```

```python
basket=['호박','감자','소주','맥주','피자']

inp = input('꺼내고 싶은 물픔을 입력하세요')

basket.remove(inp)

print(f'바구니에 소주를 꺼내어 {basket}"가 남았습니다')
```

```python
basket=['호박', '감자', '맥주', '피자']
inp1= input("추가할 물건 이름을 적어주세요")

inp2= input("어느 물품의 뒤에 넣어둘까요")
a=basket.index(inp2)+1

basket.insert(a,inp1)


print(basket)
```
