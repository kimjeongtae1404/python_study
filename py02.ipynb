## 반복문의 이해

파이썬의 반복문은 같은 코드를 여러 번 실행할 때 사용합니다.

`for` 반복문은 반복할 횟수가 정해져 있거나, 리스트·문자열·range처럼 순회할 수 있는 객체를 한 번씩 꺼내 처리할 때 자주 사용합니다.

```python
for i in range(3):
    print(i)
```

`while` 반복문은 조건이 참인 동안 계속 반복합니다. 조건이 언젠가 거짓이 되도록 설계해야 하며, 잘못하면 무한 반복에 빠질 수 있습니다.

```python
count = 0
while count < 3:
    print(count)
    count += 1
```

반복문을 활용하면 여러 데이터를 쉽게 처리할 수 있고, 장바구니 만들기, 주사위 시뮬레이션, 점수 처리처럼 실습 문제에 자주 등장합니다.
